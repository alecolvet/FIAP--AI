```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python

data_1 = np.random.multivariate_normal([1,1],[[1,0],[0,1]],1000)
data_2 = np.random.multivariate_normal([7,2],[[1,0.5],[0.5,1]],1000)
data_3 = np.random.multivariate_normal([4,8],[[1,-0.5],[-0.5,1]],1000)
data = np.concatenate((data_1, data_2, data_3), axis = 0)
plt.scatter(data[:,0],data[:,1])

plt.show()
```

```python
scores = {}
for n in range(2, 10):
    cluster = KMeans(n)
    cluster.fit(data)
    labels = cluster.labels_
    centers = cluster.cluster_centers_

    silhouette = silhouette_score(data, labels)
    scores[silhouette] = n

    plt.scatter(data[:,0], data[:,1], c = labels)
    plt.scatter(centers[:,0], centers[:,1], marker = 'x', color = 'red')
    plt.show()
plt.plot(scores.values(), scores.keys())
plt.show()

max_sil = np.max(list(scores.keys()))
n_ideal = scores[max_sil]
print(f"O número ideal de cluesters é: {n_ideal}")
```
